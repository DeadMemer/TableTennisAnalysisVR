```python
import cv2
import os

def extract_frames(video_path, output_dir, num_frames):
    """
    Estrae un numero specifico di frame dal video.
    """
    # Crea la cartella di output se non esiste
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # Apri il file video
    video = cv2.VideoCapture(video_path)

    if not video.isOpened():
        print("Errore: impossibile aprire il video.")
        return

    # Ottieni il numero totale di frame
    total_frames = int(video.get(cv2.CAP_PROP_FRAME_COUNT))
    frame_indices = [int(i * total_frames / num_frames) for i in range(num_frames)]

    extracted_frames = 0
    current_frame = 0

    while video.isOpened():
        ret, frame = video.read()
        if not ret:
            break

        if current_frame in frame_indices:
            frame_filename = os.path.join(output_dir, f"frame_{current_frame:04d}.jpg")
            cv2.imwrite(frame_filename, frame)
            extracted_frames += 1

            if extracted_frames >= num_frames:
                break

        current_frame += 1

    video.release()
    print(f"{extracted_frames} frame estratti e salvati nella cartella {output_dir}")

# Esempio di utilizzo
dir_path = "C:/Users/Edoardo/Downloads/"
video_path = dir_path + "video_quest.mp4"  # Percorso del video scaricato
output_directory = "frames"  # Directory dove salvare i frame
num_frames_to_extract = 200  # Numero di frame da estrarre

try:
    # Estrai i frame dal video locale
    extract_frames(video_path, output_directory, num_frames_to_extract)
except Exception as e:
    print(f"Errore: {e}")
```
